## Level 1: Exploring the Cosmere Library

Welcome to the ultimate Cosmere dataset! In this first section, we will download the master dataset using `kagglehub`, locate the `csv` folder, and load **`cosmere_books.csv`** to inspect Brandon Sanderson's canonical works (novels, novellas, and anthologies).

```python
import kagglehub
import pandas as pd

# Download the dataset from your Kaggle repository
path = kagglehub.dataset_download("darkacesss/cosmere-brandon-sanderson")

# Set the path pointing inside the 'csv' subfolder[cite: 1]
file_path = f"{path}/csv/cosmere_books.csv"

# Load the dataset using Pandas
df_books = pd.read_csv(file_path)

# Display dataset dimensions
print(f"Dataset Dimensions: {df_books.shape[0]} rows and {df_books.shape[1]} columns.")

# Show the first 5 records
print("First 5 Cosmere books:")
display(df_books.head(5))
```

## Level 2: Filtering and Exploring Specific Sagas

The Cosmere spans multiple worlds and major series like *The Stormlight Archive* and *Mistborn*. Let's filter our dataset to isolate specific types of publications or explore particular metrics.

```python
# Check available columns in the dataset
print("Columns in books dataset:", df_books.columns.tolist())

# Display key columns using the correct exact column names ('series_saga' and 'page_count')
display(df_books[['title', 'series_saga', 'word_count', 'page_count']].head(10))
```

## Level 3: Filtering Books by Specific Saga

Pandas allows us to filter rows dynamically using string matching. Let's filter our dataset to isolate only the books belonging to **"Mistborn Era 1"** using string containment (`str.contains`).

```python
# Filter books where 'series_saga' contains 'Mistborn Era 1'
mistborn_era1 = df_books[df_books['series_saga'].str.contains('Mistborn Era 1', na=False)]

print(f"Total books found in Mistborn Era 1: {len(mistborn_era1)}")

# Display the filtered books
display(mistborn_era1[['title', 'series_saga', 'word_count', 'page_count']])
```

## Level 4: Finding the Longest Masterpieces

Brandon Sanderson is famous for writing massive epics. Let's sort our books by **page count** in descending order to discover which publication is the longest in the entire Cosmere.

```python
# Sort books from highest to lowest page count
longest_books = df_books.sort_values(by='page_count', ascending=False)

print("📚 Top 5 Longest Books in the Cosmere:")
display(longest_books[['title', 'series_saga', 'page_count', 'word_count']].head(5))
```

## Level 5: Statistical Summary of the Cosmere Library

Let's use Pandas built-in aggregation methods to calculate total statistics, such as the total number of words written across all books, the average page count, and more.

```python
# Calculate total and average metrics
total_words = df_books['word_count'].sum()
total_pages = df_books['page_count'].sum()
avg_pages = df_books['page_count'].mean()

print(f"📊 Cosmere Library Statistics:")
print(f" - Total Words Written: {total_words:,} words")
print(f" - Total Pages Printed: {total_pages:,} pages")
print(f" - Average Page Count per Publication: {avg_pages:.1f} pages")
```
