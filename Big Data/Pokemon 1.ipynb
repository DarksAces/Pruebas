# My First Steps in Data Analysis with the Pokédex

In this notebook, we will practice fundamental data analysis operations using Python and Pandas. We will progress from simple to more complex tasks, starting with loading and viewing the data structure, to performing simple filtering and grouping.

## Exercise 1: Load the dataset and view its dimensions

The first thing every data analyst does is load the file and check how many rows (Pokémon) and columns (features) are available to work with. We will use the `kagglehub` library to download a Pokédex directly.

```python
import kagglehub
import pandas as pd

# Cargar el dataset desde tu perfil de Kaggle
file_path = "pokemon_complete.csv"
df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "darkacesss/ultimate-pokmon-pokdex-gen-1-9",
    file_path,
)

# Mostrar el número de filas y columnas
print(f"Dimensiones del DataFrame: {df.shape[0]} filas y {df.shape[1]} columnas.")

# Mostrar las primeras 3 filas para ver la estructura
display(df.head(3))
```

## Exercise 2: Filter Pokémon by elemental type

One of the most common operations in Pandas is to filter rows that meet a specific condition. We will only select Pokémon whose primary type (`type_1`) is **Fire** and see how many there are.

```python
# Filtrar los Pokémon cuyo tipo principal sea Fuego (Fire)
pokemon_fuego = df[df['primary_type'] == 'Fire']

print(f"Total de Pokémon de tipo Fuego encontrados: {len(pokemon_fuego)}")

# Mostrar los primeros 5 de la lista con su nombre y estadísticas base
display(pokemon_fuego[['pokedex_id', 'name', 'primary_type', 'base_stat_total']].head(5))
```

## Exercise 3: Sort data to find the fastest

To answer questions like "Who is number one in...?", we use the `.sort_values()` function. We will sort the Pokédex from highest to lowest maximum speed at Level 50 to see which Pokémon lead the races.

```python
# Ordenar de mayor a menor velocidad máxima a Nivel 50
pokemon_rapidos = df.sort_values(by='speed_max_lv50', ascending=False)

# Mostrar el Top 5 de los Pokémon más rápidos
print("🏆 Top 5 Pokémon más rápidos (Nivel 50):")
display(pokemon_rapidos[['pokedex_id', 'name', 'primary_type', 'speed_max_lv50']].head(5))
```

## Exercise 4: Group data by category (`groupby`)

When we want to compare groups (for example, comparing normal Pokémon versus legendary ones), we use `.groupby()`. We are going to calculate the average HP (`hp`) for each Pokémon category.

```python
# Agrupar por subcategoría y calcular la media de HP
vida_por_subcategoria = df.groupby('subcategory')['hp'].mean().reset_index()

# Redondear la media de HP a 1 decimal
vida_por_subcategoria['hp'] = vida_por_subcategoria['hp'].round(1)

# Ordenar de mayor a menor vida media para que luzca mejor
vida_por_subcategoria = vida_por_subcategoria.sort_values(by='hp', ascending=False)

# Renombrar columnas para que quede claro
vida_por_subcategoria.columns = ['Subcategoría', 'Vida Media (HP)']

print("❤️ Vida media por subcategoría de Pokémon (¡Incluyendo Pseudo-Legendarios!):")
display(vida_por_subcategoria)
```

## Exercise 5: Health Points (HP) Analysis by Elemental Type

In this exercise, we will use more powerful groupings to answer two key questions about Pokémon's health (`hp`):
1. What is the **average HP** of Pokémon by their primary type (`primary_type`), ordered from highest to lowest?
2. Which **individual Pokémon stands out the most** by having the highest HP within each elemental type?

```python
# 1. Calcular la vida media por tipo principal, redondearla y ordenarla
vida_por_tipo = df.groupby('primary_type')['hp'].mean().reset_index()
vida_por_tipo['hp'] = vida_por_tipo['hp'].round(1)
vida_por_tipo = vida_por_tipo.sort_values(by='hp', ascending=False)

print("🛡️ Vida media (HP) por Tipo Principal:")
display(vida_por_tipo)

print("\n" + "="*60 + "\n")

# 2. Encontrar el Pokémon con más HP de cada tipo principal
# Buscamos el índice del valor máximo de HP para cada grupo de tipo
idx_max_hp = df.groupby('primary_type')['hp'].idxmax()
pokemon_mas_vida_por_tipo = df.loc[idx_max_hp, ['primary_type', 'name', 'hp', 'base_stat_total']]
pokemon_mas_vida_por_tipo = pokemon_mas_vida_por_tipo.sort_values(by='hp', ascending=False)

print("👑 Pokémon con más Puntos de Salud (HP) por cada Tipo Principal:")
display(pokemon_mas_vida_por_tipo)
```
